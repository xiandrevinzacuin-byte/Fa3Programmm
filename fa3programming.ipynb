```python
score_str = input("Enter student's exam score: ")

try:
    score = int(score_str)
    if score >= 75:
        print("Passed")
    else:
        print("Failed")
except ValueError:
    print("Invalid input. Please enter a valid number for the score.")
```

```python
student_name = input("Enter student's name: ")
score_input = input(f"Enter {student_name}'s score: ")

try:
    score = int(score_input)
    if 90 <= score <= 100:
        grade = 'A'
    elif 80 <= score <= 89:
        grade = 'B'
    elif 70 <= score <= 79:
        grade = 'C'
    elif score < 70:
        grade = 'F'
    else:
        grade = 'Invalid Score (Score must be between 0 and 100)'

    print(f"\n--- Grading Result ---")
    print(f"Name: {student_name}")
    print(f"Score: {score}")
    print(f"Grade: {grade}")
    print("----------------------")

except ValueError:
    print("Invalid input. Please enter a valid number for the score.")
```
